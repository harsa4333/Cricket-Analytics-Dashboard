# Load Datasets 

```python
import pandas as pd
import numpy as np

bat = pd.read_csv(r"C:\Users\shiv_\OneDrive\Desktop\Cricket Analytics & Player Performance Intelligence Dashboard\Data\Processed\T20_Batting_Cleaned.csv")
bowl = pd.read_csv(r"C:\Users\shiv_\OneDrive\Desktop\Cricket Analytics & Player Performance Intelligence Dashboard\Data\Processed\Bowling_t20_Cleaned.csv")
field = pd.read_csv(r"C:\Users\shiv_\OneDrive\Desktop\Cricket Analytics & Player Performance Intelligence Dashboard\Data\Processed\Fielding_t20_Cleaned.csv")

bat = bat.dropna(subset=['Player'])
bowl = bowl.dropna(subset=['Player'])
field = field.dropna(subset=['Player'])
print("Shape of Batting data:", bat.shape)
print("Shape of Bowling data:", bowl.shape)
print("Shape of Fielding data:", field.shape)
```

# Merge 

```python
merged_t20 = bat.merge(
    bowl,
    on='Player',
    how='outer',
    suffixes=('_bat', '_bowl')
)
print(merged_t20.shape)
```

```python
merged_t20 = merged_t20.merge(
    field,
    on='Player',
    how='outer',
    suffixes=('', '_field')
)
print(merged_t20.shape)
```

```python
merged_t20['has_batting'] = merged_t20['Mat_bat'].notna()
merged_t20['has_bowling'] = merged_t20['Mat_bowl'].notna()

merged_t20['player_role_hint'] = 'Unknown'
merged_t20.loc[merged_t20['has_batting'] & ~merged_t20['has_bowling'], 'player_role_hint'] = 'Batsman'
merged_t20.loc[~merged_t20['has_batting'] & merged_t20['has_bowling'], 'player_role_hint'] = 'Bowler'
merged_t20.loc[merged_t20['has_batting'] & merged_t20['has_bowling'], 'player_role_hint'] = 'All-rounder'

print(merged_t20['player_role_hint'].value_counts())
```
