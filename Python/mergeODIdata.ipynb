```python
import pandas as pd
import numpy as np

bat = pd.read_csv(r"C:\Users\shiv_\OneDrive\Desktop\Cricket Analytics & Player Performance Intelligence Dashboard\Data\Processed\ODI_Batting_Cleaned.csv")
bowl = pd.read_csv(r"C:\Users\shiv_\OneDrive\Desktop\Cricket Analytics & Player Performance Intelligence Dashboard\Data\Processed\Bowling_ODI_Cleaned.csv")
field = pd.read_csv(r"C:\Users\shiv_\OneDrive\Desktop\Cricket Analytics & Player Performance Intelligence Dashboard\Data\Processed\Fielding_ODI_Cleaned.csv")

bat = bat.dropna(subset=['Player'])
bowl = bowl.dropna(subset=['Player'])
field = field.dropna(subset=['Player'])
print("Shape of Batting data:", bat.shape)
print("Shape of Bowling data:", bowl.shape)
print("Shape of Fielding data:", field.shape)
```
