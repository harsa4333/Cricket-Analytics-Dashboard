```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)

df = pd.read_csv(r"C:\Users\shiv_\OneDrive\Desktop\Cricket Analytics & Player Performance Intelligence Dashboard\Data\Processed\ODI_Player_Profile.csv")
print(df.shape)
print(df['Mat_bat'].describe())
```

```python
for threshold in [10, 20, 30, 50]:
    count = (df['Mat_bat'] >= threshold).sum()
    print(f"Mat_bat >= {threshold}: {count} players remain")
    
qualified = df[df['Mat_bat'] >= 20].copy()
print(qualified.shape)
```

```python
for col in ['Ave_bat', 'SR_bat', 'Ave_bowl', 'SR_bowl']:
    print(f"=== {col} ===")
    print("Skewness:", qualified[col].skew())
    print("Kurtosis:", qualified[col].kurt())
    print("Non-null count:", qualified[col].notna().sum())
    print()
```
